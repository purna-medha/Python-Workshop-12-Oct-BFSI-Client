## Exploration 1 — build and observe the mechanism

### 1. Represent a product

Programs act on representations of facts. Names connect those values to the problem.

**Before running:** Predict the result or state change.

```python
product_id = "1"
product_name = "USB-C Cable"
unit_price_paise = 49900
quantity = 2
print(product_id, product_name, unit_price_paise, quantity)
```

**Change and explain:** Change quantity to 3; rerun this cell.

### 2. Inspect the type of a value

A value’s type determines its available operations. Prices count paise; weight is an approximate measurement.

**Before running:** Predict the result or state change.

```python
shipping_weight_kg = 0.08
is_available = True
description = None
print(type(unit_price_paise).__name__)
print(type(shipping_weight_kg).__name__)
print(type(is_available).__name__)
print(description is None)
```

**Change and explain:** Replace None with an empty string. Does the final comparison stay true?

<details>
<summary>Reasoned expectation — reveal after predicting</summary>

int, float, bool, True; an empty string is a value distinct from absence.

</details>

### 3. Interpret incoming text

Text does not become numeric merely because it looks numeric. Conversion makes that decision explicit.

**Before running:** Predict the result or state change.

```python
quantity_text = "2"
quantity = int(quantity_text)
print(type(quantity_text).__name__, type(quantity).__name__)
print(quantity * unit_price_paise)
```

**Change and explain:** Change the text to "3" and rerun.

<details>
<summary>Reasoned expectation — reveal after predicting</summary>

The types are str and int; two cables total 99800 paise.

</details>

### 4. Evaluate and assign a result

Assignment retains the result of an evaluation. Later changes to an input do not recompute that result.

**Before running:** Predict the result or state change.

```python
quantity = 2
subtotal_paise = unit_price_paise * quantity
quantity = 3
print(subtotal_paise)
print(unit_price_paise * quantity)
```

**Change and explain:** Predict both numbers before executing.

<details>
<summary>Reasoned expectation — reveal after predicting</summary>

99800 was retained; the new expression evaluates to 149700.

</details>

### 5. Separate calculation from display

Keep stored money in integer paise. Formatting changes the presentation, not the amount.

**Before running:** Predict the result or state change.

```python
amount_paise = 49905
whole_rupees = amount_paise // 100
remaining_paise = amount_paise % 100
label = f"₹{whole_rupees:,}.{remaining_paise:02d}"
print(amount_paise, label)
```

**Change and explain:** Change amount_paise to 99800.

<details>
<summary>Reasoned expectation — reveal after predicting</summary>

49905 and ₹499.05 represent the same amount in different forms.

</details>

### 6. Prepare text for a comparison

Normalize text for matching while keeping the original label for display.

**Before running:** Predict the result or state change.

```python
product_name = "USB-C Cable"
query = "  CABLE  "
normalized_query = query.strip().casefold()
print(normalized_query)
print(normalized_query in product_name.casefold())
print(product_name)
```

**Change and explain:** Try a query that does not occur in the name.

<details>
<summary>Reasoned expectation — reveal after predicting</summary>

The normalized query is cable; membership is True; the displayed name is unchanged.

</details>

## Exploration 2 — change a condition and inspect the consequence

### 7. Choose a path

A condition selects behavior. Indentation states which instructions belong to each path.

**Before running:** Predict the result or state change.

```python
quantity = 0
if quantity > 0:
    message = "Ready to add"
else:
    message = "Choose a positive quantity"
print(message)
```

**Change and explain:** Try quantity 1, then -1.

<details>
<summary>Reasoned expectation — reveal after predicting</summary>

Zero takes the rejection-message path; this example does not mutate a cart.

</details>

### 8. Name a reusable calculation

A function exposes an operation through inputs and a returned result. Callers can reuse that result.

**Before running:** Predict the result or state change.

```python
def line_total(unit_price_paise, quantity):
    return unit_price_paise * quantity

print(line_total(49900, 2))
print(line_total(89900, 1))
```

**Change and explain:** Call the same function with price 0 and quantity 2.

<details>
<summary>Reasoned expectation — reveal after predicting</summary>

99800 and 89900; a zero-priced product contributes zero.

</details>

### 9. Distinguish printing from returning

Displaying a value is a side effect. It does not supply that value to the caller.

**Before running:** Predict the result or state change.

```python
def show_total(price_paise, quantity):
    print(price_paise * quantity)

result = show_total(49900, 2)
print("Returned value:", result)
```

**Change and explain:** Replace print with return inside the function and rerun.

<details>
<summary>Reasoned expectation — reveal after predicting</summary>

The amount is displayed; the function returns None until it has an explicit return.

</details>

### 10. Compose small functions

An application combines operations through their results. Presentation can reuse a calculation without owning it.

**Before running:** Predict the result or state change.

```python
def money(amount_paise):
    return f"₹{amount_paise // 100:,}.{amount_paise % 100:02d}"

def product_label(name, price_paise):
    return f"{name}: {money(price_paise)}"

print(product_label("USB-C Cable", 49900))
print(money(line_total(49900, 2)))
```

**Change and explain:** Add a bottle label using the same functions.

<details>
<summary>Reasoned expectation — reveal after predicting</summary>

A ₹499.00 product label and a ₹998.00 line-total label.

</details>

### 11. Observe a conversion failure

A failure exposes an assumption about input. Read its type and message before choosing a repair.

**Before running:** Predict the result or state change.

```python
quantity_text = "two"
# Supplied handling keeps the demonstration running; M2 explains try/except.
try:
    quantity = int(quantity_text)
except ValueError as error:
    print(type(error).__name__)
    print(error)
```

**Change and explain:** Identify which value violates the conversion assumption.

<details>
<summary>Reasoned expectation — reveal after predicting</summary>

ValueError: the text is not a whole-number representation.

</details>

### 12. Repair the input and reuse the operation

Repair the failed assumption, then repeat the same operation. Conversion and business permission are different decisions.

**Before running:** Predict the result or state change.

```python
quantity_text = "2"
quantity = int(quantity_text)
print(money(line_total(49900, quantity)))
print("Negative text still converts:", int("-2"))
```

**Change and explain:** Explain why converting -2 does not mean the shop should accept it.

<details>
<summary>Reasoned expectation — reveal after predicting</summary>

The repaired amount is ₹998.00. M2 adds explicit quantity rules.

</details>
